```python
#4

car_sales = pd.read_csv('car-sales.csv')

print(car_sales.head())
print(car_sales.info())
print(car_sales.describe())
print(car_sales.columns)
print(car_sales.shape)
```

```python
#5

car_sales['Price'] = car_sales['Price'].astype(str).str.replace(r'[$,]', '', regex=True)
car_sales['Price'] = pd.to_numeric(car_sales['Price'], errors='coerce').fillna(0).astype('int64')

print(car_sales.head())
```

```python
#6

car_sales['Total Sales'] = car_sales['Price'].cumsum()
print(car_sales.head())
```

```python
#7

today = date.today()
date_range = pd.date_range(start=today, periods=len(car_sales))

car_sales['Sale Date'] = date_range

print(car_sales.head())
```

```python
#8

plt.plot(car_sales['Price'],car_sales['Total Sales'])
plt.xlabel("Price")
plt.ylabel("Total Sales")
plt.title("Price vs Total Sales")
plt.show()
```

```python
#9

plt.scatter(car_sales['Odometer (KM)'], car_sales['Price'])
plt.xlabel("Odometer (KM)")
plt.ylabel("Price")
plt.title("Odometer vs Price")
plt.show()
```

```python
#10

plt.bar(car_sales['Make'], car_sales['Odometer (KM)'])
plt.xlabel("Make")
plt.ylabel("Odometer (KM)")
plt.title("Make vs Odometer (KM)")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
#11

plt.hist(car_sales['Make'])
plt.xlabel("Make")
plt.ylabel("Frequency")
plt.title("Histogram of Car Make")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
#12

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# Plot 8
axes[0, 0].plot(car_sales['Price'], car_sales['Total Sales'])
axes[0, 0].set_xlabel("Price")
axes[0, 0].set_ylabel("Total Sales")
axes[0, 0].set_title("Price vs Total Sales")

# Plot 9
axes[0, 1].scatter(car_sales['Odometer (KM)'], car_sales['Price'])
axes[0, 1].set_xlabel("Odometer (KM)")
axes[0, 1].set_ylabel("Price")
axes[0, 1].set_title("Odometer vs Price")

# Plot 10
axes[1, 0].bar(car_sales['Make'], car_sales['Odometer (KM)'])
axes[1, 0].set_xlabel("Make")
axes[1, 0].set_ylabel("Odometer (KM)")
axes[1, 0].set_title("Make vs Odometer (KM)")
axes[1, 0].tick_params(axis='x', rotation=45, labelbottom=True)


# Plot 11
axes[1, 1].hist(car_sales['Make'])
axes[1, 1].set_xlabel("Make")
axes[1, 1].set_ylabel("Frequency")
axes[1, 1].set_title("Histogram of Car Make")
axes[1, 1].tick_params(axis='x', rotation=45, labelbottom=True)


plt.tight_layout()
plt.show()
```

```python
#13

plt.figure(figsize=(8, 6))
sns.boxplot(x='Colour', y='Total Sales', data=car_sales, hue='Colour')
plt.xlabel('Colour')
plt.ylabel('Total Sales')
plt.title('Box Plot of Colour vs. Total Sales')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
#14

plt.figure(figsize=(10, 6))
sns.lmplot(y='Odometer (KM)', x='Total Sales', hue='Make', data=car_sales, ci=None)
plt.xlabel('Odometer (KM)')
plt.ylabel('Total Sales')
plt.title('Total Sales vs. Odometer (KM) by Car Make')
plt.show()
```

```python
#15

sns.pairplot(car_sales)
plt.show()
```

```python
#16

correlation_matrix = car_sales[['Total Sales', 'Price', 'Odometer (KM)']].corr(method='pearson')

plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Pearson Correlation Heatmap')
plt.show()

print("Correlation Interpretation:")
for col1 in correlation_matrix.columns:
    for col2 in correlation_matrix.columns:
      if col1 != col2:
        correlation_value = correlation_matrix.loc[col1, col2]
        print(f"The correlation between '{col1}' and '{col2}' is {correlation_value:.2f}.")
        if abs(correlation_value) > 0.5:
            print(f"This indicates a strong relationship.")
        elif abs(correlation_value) > 0.3:
          print(f"This indicates a moderate relationship.")
        else:
            print(f"This indicates a weak relationship.")

```
